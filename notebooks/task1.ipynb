```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_excel(r"C:\Users\nmark\Code\BA_Forage\ba-forage-data-science\data\flight_data.xlsx")
df.info()
df.head()
```

```python
tier_cols = ["TIER1_ELIGIBLE_PAX", "TIER2_ELIGIBLE_PAX", "TIER3_ELIGIBLE_PAX"]
total_pax = df["FIRST_CLASS_SEATS"] + df["BUSINESS_CLASS_SEATS"] + df["ECONOMY_SEATS"]

eligible_sum = df.groupby(["HAUL", "TIME_OF_DAY"])[tier_cols].sum()
total_sum = total_pax.groupby([df["HAUL"], df["TIME_OF_DAY"]]).sum()

summary = eligible_sum.div(total_sum, axis=0) * 100
summary.columns = ["TIER1_PCT", "TIER2_PCT", "TIER3_PCT"]
summary.reset_index()
```
